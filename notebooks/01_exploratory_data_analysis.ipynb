# Rajasthan Royals - SupeRR Selector Season 2
## Exploratory Data Analysis & Uncapped Talent Scouting

This notebook ingests the domestic player data pack, engineers phase-specific cricket analytics metrics, and evaluates players against Rajasthan Royals tactical archetypes.

```python
import sys
import os
sys.path.append('..')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from src.models.talent_ranker import rank_uncapped_players
from src.config import RR_TACTICAL_PRIORITIES, BENCHMARKS

sns.set_theme(style="whitegrid")
```

### 1. Load Data Pack (Raw or Sample)

```python
raw_path = '../data/raw/'
csv_files = [os.path.join(raw_path, f) for f in os.listdir(raw_path) if f.endswith(('.csv', '.xlsx'))] if os.path.exists(raw_path) else []

if csv_files:
    print(f"Loading official data pack: {csv_files[0]}")
    df_raw = pd.read_excel(csv_files[0]) if csv_files[0].endswith('.xlsx') else pd.read_csv(csv_files[0])
else:
    print("Loading sample benchmark dataset...")
    df_raw = pd.read_csv('../data/sample/uncapped_players_sample.csv')

df_raw.head()
```

### 2. Run Ranking & Tactical Archetype Engine

```python
df_ranked = rank_uncapped_players(df_raw)
top10 = df_ranked.head(10)

top10[['rank', 'name', 'age', 'role', 'domestic_league', 'royals_composite_score', 'rr_tactical_archetype', 'estimated_auction_bracket']]
```

### 3. Visualize Top 10 Prospects

```python
plt.figure(figsize=(12, 6))
sns.barplot(data=top10, x='royals_composite_score', y='name', hue='rr_tactical_archetype', palette='viridis', dodge=False)
plt.title('Top 10 Youngest Uncapped Targets for Rajasthan Royals (IPL 2027)', fontsize=14, fontweight='bold')
plt.xlabel('Royals Composite Score')
plt.ylabel('Player')
plt.tight_layout()
plt.show()
```
